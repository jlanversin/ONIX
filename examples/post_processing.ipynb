```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
def plot_nuc(i, nuc):
    core_nuc = core_den.loc[nuc].to_frame()
    loop_nuc = loop_den.loc[nuc].to_frame()
    blanket_nuc = blanket_den.loc[nuc].to_frame()
    reprocessing_plant_nuc = reprocessing_plant_den.loc[nuc].to_frame()

    core_nuc.reset_index(inplace=True)
    loop_nuc.reset_index(inplace=True)
    blanket_nuc.reset_index(inplace=True)
    reprocessing_plant_nuc.reset_index(inplace=True)

    plt.figure(i+1)
    x = time
    y = core_nuc.iloc[0:-1, 1].values.tolist()
    y[0] = float(y[0])
    y = [1e24*item for item in y]
    plt.plot(x, y, linestyle='-', color='b', label="Core")

    z = loop_nuc.iloc[0:-1, 1].values.tolist()
    z[0] = float(z[0])
    z = [1e24*item for item in z]
    plt.plot(x, z, linestyle='-', color='r', label="Loop")

    a = blanket_nuc.iloc[0:-1, 1].values.tolist()
    a[0] = float(a[0])
    a = [1e24*item for item in a]
    plt.plot(x, a, linestyle='-', color='g', label="Blanket")

    b = reprocessing_plant_nuc.iloc[0:-1, 1].values.tolist()
    b[0] = float(b[0])
    b = [1e24*item for item in b]
    plt.plot(x, b, linestyle='-', color='k', label="Reprocessing Plant")

    plt.xlabel("Time [Years]")
    plt.ylabel("Density")
    plt.title(nuc)
    #plt.yscale('log')
    #plt.xlim(0, 50)

    # Show grid
    plt.grid(True)

    # Display legend
    plt.legend()
    plt.show()
    plt.close()
```
